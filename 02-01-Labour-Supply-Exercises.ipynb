# Предлагане на труд {#sec-labour-supply-exercises}


## Задача 2-3 {#sec-ex-2-3}

Том печели по \$15 на час за работа до 40 часа седмично. За всеки час над 40-ия му се заплаща по \$30 на час. Том е облаган с 20 процента данъчна ставка и плаща по \$4 на час за разходи за грижи за дете за всеки час, в който работи. Том получава по \$80 на седмица под формата на плащания за издръжка на дете. В седмицата има 110 часа (време, в което не се спи). Начертайте седмичната бюджетна линия на Том.

## Задача 2-4 {#sec-ex-2-4}

Синди извлича полезност от потреблението $C$ и свободното време (отдиха) $L$. Максималното свободно време, което тя може да консумира през всяка дадена седмица, е 110 часа. Нейната функция на полезност е $U(C, L) = C \cdot L$. Тази функционална форма предполага, че пределната норма на заместване (MRS) на Синди е $C/L$. Синди получава \$660 всяка седмица от своята прабаба – независимо от това колко работи. Каква е резервационната заплата на Синди?

## Задача 2-6 {#sec-ex-2-6}

Предпочитанията на Шели по отношение на потреблението и свободното време могат да бъдат изразени чрез функцията:

$$
U(C, L) = (C - 100)(L - 40)
$$

Тази функция на полезност предполага, че пределната (маргиналната) полезност на свободното време за Шели е $C - 100$, а нейната пределна полезност на потреблението е $L - 40$.

В седмицата има 110 часа (време, в което не се спи), които тя може да разпредели между работа и свободно време. Шели печели по \$10 на час след облагане с данъци. Тя също така получава социални помощи в размер на \$320 всяка седмица, независимо от това колко работи.

а) Начертайте бюджетната линия на Шели.

б) Каква е пределната степен на замяна (MRS) на Шели, когато $L = 100$ и тя се намира върху своята бюджетна линия?

в) Каква е резервационната работна заплата на Шели?

г) Намерете оптималното количество потребление и свободно време за Шели.

## Задача 2-8 {#sec-ex-2-8}

- През 1999 г. 4860 получатели на помощи по програмата TANF (Temporary Assistance for Needy Families) са били попитани колко часа са работили през предходната седмица.
- През 2000 г. 4392 от тези получатели отново са били обект на същите правила на TANF и отново са били попитани за отработените им часове през предходната седмица.
- Останалите 468 лица са били разпределени на случаен принцип към експеримент с т.нар. „негативен подоходен данък“ (NIT), който предоставя финансови стимули за работа на получателите на социални помощи и ги подчинява на своите правила. Подобно на другата група, те също са били попитани за работните им часове през предходната седмица. Данните от експеримента се съдържат в таблицата по-долу.

| | Общ брой получатели | Работили през седмицата на проучването, 1999 | Работили през седмицата на проучването, 2000 | Общо часове труд, 1999 | Общо часове труд, 2000 |
|---|---:|---:|---:|---:|---:|
| TANF | 4 392 | 1 217 | 1 568 | 15 578 | 20 698 |
| NIT | 468 | 131 | 213 | 1 638 | 2 535 |
| Общо | 4 860 | 1 348 | 1 781 | 17 216 | 23 233 |

а) Какъв ефект е имал експериментът с NIT върху коефициента на заетост на получателите на обществена помощ? Съставете стандартна таблица за метода „разлика в разликите“ (difference-in-differences, DiD), за да подкрепите отговора си.

б) Какъв ефект е имал експериментът с NIT върху седмичните отработени часове на тези получатели на обществена помощ, които са имали положителен брой работни часове през седмицата на проучването? Съставете стандартна таблица за метода „разлика в разликите“, за да подкрепите отговора си.

## Задача 2-14 {#sec-ex-2-14}

Разгледайте случай на лице, което може да работи до 80 часа на седмица при брутна заплата от \$20 на час, но плаща постоянен данък върху възнаграждението в размер на 20%. При тези условия работникът максимизира своята полезност, като избира да работи 50 часа седмично. Правителството предлага въвеждането на отрицателен подоходен данък, при който всеки получава по \$300 на седмица (трансфер), като всеки може да допълва дохода си чрез работа. За да се финансира този отрицателен подоходен данък, данъкът върху възнагражденията се увеличава на 50%.

а) На една графика начертайте първоначалното бюджетно ограничение на работника и бюджетното му ограничение при системата с отрицателен подоходен данък.

б) Покажете, че работникът ще избере да работи по-малко часове, ако се приеме отрицателният подоходен данък.

в) Ще бъде ли полезността на работника по-висока при системата с отрицателен подоходен данък?
