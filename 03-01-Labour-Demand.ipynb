# Labour Demand

In @sec-labour-supply we discussed the basic labour supply model. Here we turn to the other side of the market: firmst that produce and sell goods and hire workers.

The model development here follows quite closely the model from the labour supply section as the concept of indifference curves transfers almost in a one-to-one way to
the production decision of firms (XXX, wording, check).

We assume that a firm faces a fixed market price (a perfect competition setting) and that the level of production of the firm is small enough compared to the
whole market so that it does not influence the sales price which we denote by $p$ (dollars per unit of goods).

We describe the technology of the firm by a production function that we call $F(Е, L)$ which takes only two inputs for now: amount of capital (factories, trucks, other machines, land) and the number of work-hours (L).
We will also make a couple of (resonable) assumptions about the technology, namely deminishing marginal returns to both capital and labour. In other words we assume that holding capital fixed, one additional unit of work-hours
produces less and less additional product the more labour is employed. We assume the same for capital: holding the amount of work-hours fixed, increasing the capital employed brings less and less additional product. Finally, we assume that more
of at least one of the inputs always increases production (the production surface is not declining anywhere): meaning that hiring an additional worker or installing an additional processing machine does not reduce the level of production.


$$
q = f(E, K) \quad \text{measured in e.g. kg}
$$

Given these assumptions we can derive iso-production curves in the same way as we did in the case of the utility function.

The firms pay the hourly wage $w$ (also determined externally by the market) per work-hours and also pays a rent $r$ (TODO: interest?) for the capital used (think renting offices, machines, etc.).

## The Profit Function

The unit of the production function (q) is a quantity of goods (e.g. kg, tons, litres, etc.) which the firm sells on the market to obtain the revenue $pq$

$$
\text{Profit} = \underbrace{pq}_{\text{Revenue}} - \underbrace{(wE + rK)}_{\text{Costs}}
$$

Next we assume what drives the behaviour of the firm: it maximizes its profit. As the firm takes the market price of its output
as fixed, the only thing that it can adjust to optimize the production level (and therefore the labour employment decision) is
to minimise the costs or production.

For the maximization we can distinguish between two time horizons

- The short run: the firm cannot change the level of capital (think about building a new factory)
- The long run: the optimization decision also includes changes to the capital stock (XXX, wording, check)



## Short Run

In the short run we treat $K$ as fixed. Let's see how the optimal employment strategy of the firm looks like. We are looking for the optimal 
production decision, in other words the maximal profit. First we find the derivative of the profit function with respect to $E$ (remember, $K$ is fixed so it drops out when calculating the derivative),
set it to zero and solve.

$$
\frac{d}{dE}\text{Profit} = p\frac{dq}{dE} - w = 0
$$

Rearranging we get

$$
p \,dq = w \,dE
$$


$$
\frac{dq}{dE} = \frac{w}{p}
$${#eq-demand-opt-short-run-ratio}

Let's translate equation @eq-demand-opt-short-run in plain words. It tells us that we can find the best output level at a point where the additional output per additional work-hour 
just equals the price for the additinal work-hour (the wage rage) divided by the sales price of the product (XXX, finish). In the optimum, the costs ($w$) of paying for one additional hour 
of work is just equal to the revenue obtained by selling the additional producted produced with that additional (XXX, check repetitions) hour.


In the long run the firm can also vary the capital stock ($K$) (XXX, check stock?) and so the maximization is done with respect to both variables.

$$
\begin{align*}
\frac{\partial}{\partial E}\text{Profit} = pf_E - w = 0 \\
\frac{\partial}{\partial K}\text{Profit} = pf_K - r = 0 \\
\end{align*}
$$

The two optimality conditions state that the first should keep expanding both work-hours (hiring) and capital until the marginal prices (for one additional unit) of both production
factors just equal the value of the marginal products.

We can take the ratio of both equations:

$$
\frac{\cancel{p} f_E}{\cancel{p} f_K} = \frac{w}{r}
$$
