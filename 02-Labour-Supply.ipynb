# A Simple Labour Supply Model

## A Brief Overview of the Main Concepts

The first model we encounter in the course presents and idealized and very simplified view of the behaviour of workers. In this small world, people work in order
to generate income that they spend on consumption. There are two sources of income: labour income that earns the wage rate ($w$) per hour and the worker must decide how many hours ($L$) he or she
will work (at that wage rate). People have a limited time budget ($T$ hours) per week and working would reduce the time available ($L$) for leisure activities.


$$
f(C, L)
$$

with respect to consumption (C) of goods and services and leisure time (L) under a budget constraint.

$$
w(T - L) + V = C
$$

where C is the worker's consumption, T is the total available time to allocate to work and leisure in a week (i.e. after subtracting sleep time, ect.), $w$ is the hourly wage rate.

$$
\underset{C,L}{\max} f(C,L) \quad \text{s.t.} \quad w(T - L) + V - C = 0
$$

Under some assumptions about the nature of the utility function (XXX, list these), the maximization problem can be solved using the Lagrangian:

$$
\mathcal{L}(C,L,\lambda) = f(C, L) + \lambda(wT - wL + V - C)
$$

The first order conditions are

$$
\begin{align*}
\frac{\partial}{\partial C}\mathcal{L} & = f_{C} - \lambda = 0 \\
\frac{\partial}{\partial L}\mathcal{L} & = f_{L} - \lambda w = 0 \\
\frac{\partial}{\partial \lambda}\mathcal{L} & = wT - wL + V - C = 0 \\
\end{align*}
$$

where $f_C, f_L$ are the partial derivatives of the utility functions with respect to $C$ and $L$. We call these partial derivatives the marginal utility of consumption and leisure, respectively (XXX, rephrase).


## Interpretation of the Optimality Condition

(XXX, introduce the marginal utilities before that)

From the first two equations we see the optimality condition:

$$
\frac{f_L}{f_C} = w
$${#eq-supply-optimality-condition}

Thus, at the optimum the ratio of the marginal utility of leisure and the marginal utility of consumptino equal the wage rate (XXX, deeper interpretation needed).

Also note the interpretation of $\lambda$. To see this, it is convenient to rearrange the budget constraint so that the total income is on the lefthand side:

$$
wL + C = wT + V
$$

so that it represents the total income (labour plus non-labour) at a fixed wage rate. The shadow price of the constraints show the increase in total utility in relation to a small increase in the
total income (lefthand side).

Consider a point (C, L) combination where the ratios of the marginal utilities in @eq-supply-optimality-condition is is larger than the wage rate. This means that the worker
can reduce the hours worked, forego consumption equal to $w$ (XXX, check condition) but he/she will be better-off, becase the utility gained by one more hour of leisure outweights
the reduced consumption.

The same line of reasoning rules out optimality for points where the ratio in @eq-supply-optimality-condition is less than the wage rate: he/she can work more, sactifice some
leisure time but be ultimately better off.


## Visualizations of the Utility Function

The utility function maps each combination of C and L (consumption and leisure) to a point in $\mathbb{R}$. One way to plot it is to show it in 3D. Assuming that persons are happier consuming more goods and having more leisure time, the function must be increasing in both $C$ and $L$ directions. Another common feature of the utility function is that it grows more slowly at high levels of consumption and leisure (deminishing marginal utility), reflecting the idea (assumption) that
people value leisure more when they only have a little of it (XXX, rephrase) and the same applies to consumption.

## Slope of the iso-utility curves

Each curve describes combinations of C and L, corresponding to a fixed level of utility, say $u$:

$$
f(C, L) = u
$${#eq-supply-iso-condition}

Among other things @eq-supply-iso-condition means that when we travel along one and the same iso-line the total utility does not change. The total differential of $f(C,L)$ along a single iso-utility line must be zero
(meaning that the function does not change when we move along the line):

$$
df(C, L) = \frac{\partial f(C, L)}{\partial C}dC + \frac{\partial f(C, L)}{\partial L}dL = 0
$$

To simplify it a bit, let's call $\partial f(C, L)/ \partial L = MU_L$ and $\partial f(C, L)/ \partial C = MU_C$.

$$
MU_C dC + MU_L dL = 0 \implies -\frac{MU_C}{MU_L} = \frac{dL}{dC}
$$

The last equality is important so let's write it cleanly and explain it:

$$
-\frac{MU_C}{MU_L} = \frac{dL}{dC}
$$

Consider increasing leisure by some small amount $dL$ (but keep $C$ at a fixed level). Because more leisure leads to higher utility (at an unchanged consumption), the consumption must decrease somewhat so that
we don't wander off the fixed iso-curve. The increase in utility due to more leisure ($MU_L dL$) must be offset by a decrease in utility due to lower consumption $MU_C dC$.


## The Budeget Constraint

In the current version of the model we assume a very simple structure of the budget constraint of the workers. To consume goods they must spend cash from one of two sources (XXX, this is repeated, check). For now we will be rather vague about the sources of non-labour income but think about renting an apartment, interest earned on savings, etc. The amount of work time is the difference between the total available time $T$ per week (for example) and the leisure time $L$.

$$
w(T - L) + V = C
$$

In the xy-plane this equation describes a straight line with a slope of $-w$ intercepting the y-axis (zero leisure) at $wT + V$

$$
C = -wL + wT + V
$$

The optimal condition states that the worker will choose (in our model) the combination C and L for which the ratio of $MU_L$ and $MU_C$ equals $w$. Note that we showed the ratio of the marginal utilities to equal
the slope of the indifference curves (with a minus sign) and that negative $w$ is the slope of the budget constraint line. Geometrically we can find the optimum to be at a point where the budget line is tangent to a iso-utility curve (at the level of the maximal attainable utility).
